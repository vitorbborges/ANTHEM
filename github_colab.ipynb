# 🛠️ GitFlow Branching Model: What You Need to Know

**GitFlow** is a branching model for Git that helps teams collaborate more smoothly by organizing different types of work — like adding new features, fixing bugs, or preparing releases — into separate branches.

For a more in-depth look at GitFlow, check out this awesome [GitFlow documentation](https://nvie.com/posts/a-successful-git-branching-model/).

Here's a quick breakdown of how it works:

---

## 🔵 `main`
- This is your **production** branch.  
- Only stable, tested, and ready-to-deploy code lives here.  
- Every commit on `main` should represent a version that could be released to users.

---

## 🟣 `develop`
- This is your **integration** branch.  
- All new features and completed work get merged into `develop` first.  
- Think of it as the "staging area" before the code is promoted to `main`.

---

## 🌿 `feature/your-feature-name`
- Used for **building new features** or doing focused work.  
- Created from `develop`, merged back into `develop` when done.  
- Naming format: `feature/login-page`, `feature/api-refactor`, etc.

---

## 📦 `release/x.y.z` (optional for bigger teams)
- Created from `develop` when you're preparing a new version.  
- Final tweaks, version bumps, or minor bug fixes happen here.  
- Merged into both `main` and `develop` afterward.

---

## 🐞 `hotfix/x.y.z`
- Created from `main` to quickly fix something in production.  
- Merged into both `main` and `develop` after the fix.

---

# 🧭 GitFlow in Practice

1. You **create a new feature branch** from `develop`
2. You **make your changes**
3. You **commit and push** your branch
4. You open a **Pull Request (PR)** to merge your feature into `develop`
5. Eventually, `develop` is merged into `main` during a release

---

## 📝 Example Branch Flow

```
main ──────●──────────────●─────────────► production
             ╲              ╲
              ╲              ╲
             hotfix/1.0.1    release/1.1 ──► (final polishing)
                       ╲
                        ╲
                       develop ──●─────●─────► ongoing development
                                 ╲
                                  ╲
                           feature/new-report
```

## 🧠 GitFlow Summary

| Branch Type   | Purpose                          | Based On   | Merges Into         |
|---------------|----------------------------------|------------|----------------------|
| `main`        | Production-ready code            | -          | -                    |
| `develop`     | Ongoing development integration  | `main`     | `main`               |
| `feature/*`   | Individual features or fixes     | `develop`  | `develop`            |
| `release/*`   | Final prep before production     | `develop`  | `main` + `develop`   |
| `hotfix/*`    | Urgent fixes to production       | `main`     | `main` + `develop`   |

Use GitHub to open a Pull Request from your `feature/` branch to `develop` once your work is done!

# 🔐 GitHub Colab Safety Setup

Welcome! This notebook is connected to a shared GitHub repository and follows a structured GitFlow-like workflow. To ensure collaborators never accidentally commit sensitive or shared configuration files, this notebook includes a **Git Safety Check** that:

- ✅ Reactivates the `post-checkout` Git hook if needed
- ✅ Reapplies `assume-unchanged` rules to protected files
- ✅ Ensures the working environment is clean and consistent

> You **don't need to run any manual Git commands** to protect key files — everything is automated below.


````python
import os

# List of files to protect
protected_files = [
    "github_colab.ipynb",  # Replace or add more as needed
]

# Path to Git hook
hook_path = ".git/hooks/post-checkout"

# Step 1: Ensure the hook is executable
if os.path.exists(hook_path):
    os.chmod(hook_path, 0o755)
    print("✅ Git hook is executable.")
else:
    print("⚠️ Git hook not found at", hook_path)

# Step 2: Manually apply assume-unchanged protection
for file in protected_files:
    if os.path.exists(file):
        os.system(f"git update-index --assume-unchanged {file}")
        print(f"🔒 Protected: {file}")
    else:
        print(f"⚠️ File not found: {file}")

# Step 3: Optional Git status summary
print("\n📍 Current Git status:")
!git branch -vv
````

## 🔄 How to Open a Pull Request on GitHub

After pushing your changes to a `feature/` branch, it’s time to **merge your work back into `develop`** (or `main` if you're on a hotfix branch).

Here’s how to do it:

### 📌 Step-by-Step: Creating a Pull Request (PR)

1. **Go to your repository** on GitHub:  
   [https://github.com/your-username/your-repo-name](https://github.com/your-username/your-repo-name)

2. GitHub will detect your recently pushed branch and display a **“Compare & pull request”** button. Click it.  
   _Or, click the "Pull requests" tab and then “New pull request”._

3. In the PR page:
   - **Base branch**: Choose `develop`  
   - **Compare branch**: Choose your `feature/your-feature-name`
   - Add a clear **title** and **description** of what your changes do

4. 📣 _Optional_: Tag reviewers if needed

5. Click **“Create pull request”**

---

### ✅ When is your PR ready to merge?

- ✅ You’ve tested your changes
- ✅ The code is clean and commented
- ✅ The commit history is meaningful
- ✅ It doesn’t include unintended files (use `git status`!)

Once approved, you or the team lead can merge the PR into `develop`.

🧼 After merging, remember to delete the feature branch to keep the repo clean (GitHub usually offers a button for this).
