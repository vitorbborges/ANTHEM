```python
import pandas as pd
import os
```

```python
df = pd.concat([pd.read_parquet(f"data/processed_data/{file}") for file in os.listdir("data/processed_data/")])
df.drop(columns=['PM25'], inplace=True, errors='ignore')
df.loc[df['average_nearby_num_lanes_50'].isna(), 'average_nearby_num_lanes_50'] = 0
df.loc[df['average_nearby_maxspeed_50'].isna(), 'average_nearby_maxspeed_50'] = 0
df.loc[df['average_nearby_streets_len_50'].isna(), 'average_nearby_streets_len_50'] = 0
df.loc[df['average_building_height_100'].isna(), 'average_building_height_100'] = 0
df.loc[df['average_building_height_200'].isna(), 'average_building_height_200'] = 0
df.dropna(inplace=True)
```

```python
df.to_parquet("data/processed_data/combined_subjects.parquet")
```
