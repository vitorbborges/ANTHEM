```python
# c) land-use areas (residential, commercial, industrial…)
tags_landuse = {"landuse": True}
gdf_landuse = features_from_bbox(BBOX, tags_landuse)
```

```python
feature_name = "close2industry_400"

df[feature_name] = df.apply(
    lambda row: is_close_to(
        features=gdf_landuse,
        point=Point(row.x, row.y),
        threshold=400,
        type_column="landuse",
        types="industrial",
    ), axis=1
).astype(int)

# add a new metadata field
docs.setdefault(feature_name, {})
docs[feature_name]["description"] = "'Dummy' variable indicating wheter the point is less than 400m away from an industrial area"
docs[feature_name]["type"] = "integer"
docs[feature_name]["range"] = {int(df[feature_name].min()), int(df[feature_name].max())}
docs[feature_name]["created_on"] = "osmnx_landuse.ipynb"
docs[feature_name]["source"] = "OSM"
docs[feature_name]["source_url"] = "https://www.openstreetmap.org/"
```
