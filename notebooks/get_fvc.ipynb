```python
import pandas as pd
import numpy as np
from itertools import product
```

```python
age = list(range(16,80))
height = [i/100 for i in range(150, 201, 1)]
sex = ['M', 'F']
ethnic = [0]

list(product(age,height))
```

```python
df = pd.DataFrame(np.array(list(product(age, height, sex, ethnic))))
df.columns = ['age', 'height', 'sex', 'ethnic']
df
```

```python
n = len(df) // 1000
for i in range(n+1):
    idx = i * 1000
    df.iloc[idx:idx+1000].to_excel(f'notebooks/fvc_data/{i}.xlsx', index=False)
```
