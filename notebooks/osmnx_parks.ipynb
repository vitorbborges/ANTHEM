```python
import osmnx as ox
from osmnx.features import features_from_bbox
import geopandas as gpd
from src.feature_building_utils import *
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import toml
```

```python
df = pd.read_parquet('data/processed_data/S1CO2-approx-coordinates.parquet')
```

```python
# Option A: completely empty the file
TOML_PATH = "documentation/feature_docs.toml"
open("documentation/feature_docs.toml", "w").close()
docs = toml.load("documentation/feature_docs.toml")
```

```python
west, south, east, north = 9.2257, 45.47162, 9.23768, 45.48537
bbox = (west, south, east, north)
```

```python
# b) parks & town squares
tags_parks = {
    "leisure": ["park", "garden"],
    "amenity": ["town_square"]
}
gdf_parks = features_from_bbox(bbox, tags_parks)
```

```python
feature_name = "inside_green"

df[feature_name] = df.apply(
    lambda row: is_inside(
        features=gdf_parks,
        point=Point(row.x, row.y),
        type_column="leisure",
        types=["park", "garden"]
    ), axis=1
).astype(int)

# add a new metadata field
docs.setdefault(feature_name, {})
docs[feature_name]["description"] = "'Dummy' variable indicating wheter the point is inside a green area or not"
docs[feature_name]["type"] = "boolean"
docs[feature_name]["range"] = {0, 1}
docs[feature_name]["created_on"] = "osmnx_parks.ipynb"
docs[feature_name]["source"] = "OSM"
docs[feature_name]["source_url"] = "https://www.openstreetmap.org/"

with open("documentation/feature_docs.toml", "w") as f:
    toml.dump(docs, f)
```

```python
feature_name = "close2park_25"

df[feature_name] = df.apply(
    lambda row: is_close_to(
        features=gdf_parks,
        point=Point(row.x, row.y),
        threshold=25,
        type_column="leisure",
        types=["park", "garden"],
    ), axis=1
).astype(int)

# add a new metadata field
docs.setdefault(feature_name, {})
docs[feature_name]["description"] = "'Dummy' variable indicating wheter the point is less than 25m away from a park or not"
docs[feature_name]["type"] = "boolean"
docs[feature_name]["range"] = {0, 1}
docs[feature_name]["created_on"] = "osmnx_parks.ipynb"
docs[feature_name]["source"] = "OSM"
docs[feature_name]["source_url"] = "https://www.openstreetmap.org/"

with open("documentation/feature_docs.toml", "w") as f:
    toml.dump(docs, f)
```
