```python
import osmnx as ox
from osmnx.features import features_from_bbox
import geopandas as gpd
from src.feature_building_utils import *
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import toml
from shapely.geometry import Point
```

```python
df = pd.read_parquet('data/processed_data/S3-approx-coordinates.parquet')
```

```python
docs = toml.load("documentation/feature_docs.toml")
```

```python
# b) parks & town squares
tags_parks = {
    "leisure": ["park", "garden"],
    "amenity": ["town_square"]
}
gdf_parks = features_from_bbox(BBOX, tags_parks)
```

```python
feature_name = "inside_green"

df[feature_name] = df.apply(
    lambda row: is_inside(
        features=gdf_parks,
        point=Point(row.x, row.y),
        type_column="leisure",
        types=["park", "garden"]
    ), axis=1
).astype(int)

# add a new metadata field
docs.setdefault(feature_name, {})
docs[feature_name]["description"] = "'Dummy' variable indicating whether the point is inside a green area or not"
docs[feature_name]["type"] = "boolean"
docs[feature_name]["range"] = {0, 1}
docs[feature_name]["created_on"] = "osmnx_parks.ipynb"
docs[feature_name]["source"] = "OSM"
docs[feature_name]["source_url"] = "https://www.openstreetmap.org/"
```

```python
feature_name = "close2park_25"

df[feature_name] = df.apply(
    lambda row: is_close_to(
        features=gdf_parks,
        point=Point(row.x, row.y),
        threshold=25,
        type_column="leisure",
        types=["park", "garden"],
    ), axis=1
).astype(int)

# add a new metadata field
docs.setdefault(feature_name, {})
docs[feature_name]["description"] = "'Dummy' variable indicating whether the point is less than 25m away from a park or not"
docs[feature_name]["type"] = "boolean"
docs[feature_name]["range"] = {0, 1}
docs[feature_name]["created_on"] = "osmnx_parks.ipynb"
docs[feature_name]["source"] = "OSM"
docs[feature_name]["source_url"] = "https://www.openstreetmap.org/"
```

```python
feature_name = "close2garden_25"

df[feature_name] = df.apply(
    lambda row: is_close_to(
        features=gdf_parks,
        point=Point(row.x, row.y),
        threshold=25,
        type_column="leisure",
        types=["garden"],
    ), axis=1
).astype(int)

# add a new metadata field
docs.setdefault(feature_name, {})
docs[feature_name]["description"] = "'Dummy' variable indicating whether the point is less than 25m away from a garden or not"
docs[feature_name]["type"] = "boolean"
docs[feature_name]["range"] = {0, 1}
docs[feature_name]["created_on"] = "osmnx_parks.ipynb"
docs[feature_name]["source"] = "OSM"
docs[feature_name]["source_url"] = "https://www.openstreetmap.org/"
```

```python
feature_name = "num_green_200"

df[feature_name] = df.apply(
    lambda row: count_nearby(
        features=gdf_parks,
        point=Point(row.x, row.y),
        threshold=200,
        type_column="leisure",
        types=["park", "garden"],
    ), axis=1
).astype(int)

# add a new metadata field
docs.setdefault(feature_name, {})
docs[feature_name]["description"] = "Number of green areas within 200m from the point"
docs[feature_name]["type"] = "integer"
docs[feature_name]["range"] = {int(df[feature_name].min()), int(df[feature_name].max())}
docs[feature_name]["created_on"] = "osmnx_parks.ipynb"
docs[feature_name]["source"] = "OSM"
docs[feature_name]["source_url"] = "https://www.openstreetmap.org/"
```
