```python
import os
import pandas as pd
from scipy.io import loadmat
```

```python
# Define the base directory where the files are stored
base_dir = "data/adapted_raw_data/"
```

```python
# Load data for one subject and one pollutant (CO2)
subject = 1
pollutant = 'CO2'

# Build paths
subj_folder_name = f"S{subject:02d}"    # e.g., S01
subj_prefix = f"S{subject}"             # e.g., S1
base_dir = "data/adapted_raw_data/"
subject_path = os.path.join(base_dir, subj_folder_name)
```

```python
# Container for all CO2-related data
co2_data = {}

# Load .mat file
mat_file = f"{subj_prefix}{pollutant}.mat"
mat_path = os.path.join(subject_path, mat_file)
if os.path.exists(mat_path):
    mat_contents = loadmat(mat_path)
    mat_vars = {k: v for k, v in mat_contents.items() if not k.startswith('__')}
    co2_data['mat'] = mat_vars
    print(f"✅ Loaded {mat_file} with fields: {list(mat_vars.keys())}")
else:
    print(f"⚠️  .mat file {mat_file} not found")
```

```python
# Extract example variables from the .mat file
co2_matrix = mat_vars.get('AB1')
if co2_matrix is not None:
    print(f"CO2 matrix shape: {co2_matrix.shape}")
    print(f"CO2 matrix preview:\n{co2_matrix[:5, :]}")
```

## Datetime values (t1 and ttt) are saved as strings

```python
time_array = mat_vars.get('t1')  # might be time values t1 and ttt
if time_array is not None:
    print(f"Time array shape: {time_array.shape}")
    print(f"Time array preview:\n{time_array[:5]}")
```

```python
pd.to_datetime(time_array, format='%H:%M:%S')
```

```python
pd.to_datetime(time_array, format='%H:%M:%S').time
```
